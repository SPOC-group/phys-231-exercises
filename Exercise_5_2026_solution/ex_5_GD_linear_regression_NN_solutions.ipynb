## Ex 3: Linear regression trained with gradient descent

We use linear regression as a binary classifier for odd/even MNIST. We call $n$ the number of training samples and $d$ the number of pixels ($d=784$). For a weight vector $w\in\mathbb R^d$, the predicted class is 
$$z_\mu=\mathrm{sign}(X_\mu w).$$

We train with the square loss
$$
\mathcal L(w)=\frac{1}{2n}\sum_{\mu=1}^n (X_\mu w-y_\mu)^2
=\frac{1}{2n}\|Xw-y\|_2^2.
$$


#### Ex 3.1:  Implementing the gradient
The gradient of the square loss is
$$
\nabla_w\mathcal L(w)=\frac{1}{n}X^T(Xw-y).
$$
Implement a function `loss_gradient(X, y, w)` that returns this gradient.

```python
### your code goes here ###
```

```python
# Solution:
def loss_gradient(X, y, w):
    return X.T@(X@w-y)/X.shape[0]
```

Your answer here: 

```python
# Solution:
n_small = 300
rng = np.random.default_rng(0)
train_order = rng.permutation(len(X_train))
X_small = X_train[train_order[:n_small]]
y_small = y_train[train_order[:n_small]]

n, d = X_small.shape

# A stable step size for the square loss.
L = np.linalg.norm(X_small, ord=2)**2/n
gamma = 1.9/L 


tmax = 100000
measure_spacing = 100

# Same random direction, different initialization norms.
direction = rng.normal(size=d)
direction /= np.linalg.norm(direction)
w_small = 1e-2*direction
w_large = 5.0*direction

times = []
train_loss_small, train_loss_large = [], []
train_acc_small, train_acc_large = [], []
norm_small, norm_large = [], []
w_small_history, w_large_history = [], []


# Helper functions that we will reuse for this exercise.
def square_loss(X, y, w):
    return 0.5*np.mean((X@w-y)**2)


def accuracy(X, y, w):
    y_pred = np.where(X@w >= 0, 1, -1)
    return np.mean(y_pred == y)



for t in range(tmax+1):
    if t<500 or t % measure_spacing == 0:
        times.append(t)

        train_loss_small.append(square_loss(X_small, y_small, w_small))
        train_loss_large.append(square_loss(X_small, y_small, w_large))
        train_acc_small.append(accuracy(X_small, y_small, w_small))
        train_acc_large.append(accuracy(X_small, y_small, w_large))

        norm_small.append(np.linalg.norm(w_small))
        norm_large.append(np.linalg.norm(w_large))
        w_small_history.append(w_small.copy())
        w_large_history.append(w_large.copy())

    if t < tmax:
        w_small -= gamma*loss_gradient(X_small, y_small, w_small)
        w_large -= gamma*loss_gradient(X_small, y_small, w_large)

plt.plot(times, train_loss_small, label="small initialization")
plt.plot(times, train_loss_large, label="large initialization")
plt.xscale("log")
plt.yscale("log")
plt.xlabel("t (steps)")
plt.ylabel("train loss")
plt.grid(alpha=0.5)
plt.legend()
plt.show()

plt.plot(times, train_acc_small, label="small initialization")
plt.plot(times, train_acc_large, label="large initialization")
plt.xlabel("t (steps)")
plt.xscale("log")
plt.ylabel("train accuracy")
plt.grid(alpha=0.5)
plt.legend()
plt.show()
```

**Solution** Both runs can reach essentially the same training loss and training accuracy. This does not imply that they converge to the same weight vector; we check this next.


#### Ex 3.3: Implicit regularization

In last week's exercise, you saw the ridge-regression analytical solution. Here $n<d$, so $X_{\rm small}^TX_{\rm small}$ is necessarily singular, so let us add a **very small** $L_2$ regularization
$$
\mathcal L_\lambda(w)=\frac{1}{2n}\left(\|Xw-y\|_2^2+\lambda\|w\|_2^2\right),
$$
for which the analytical solution is
$$
w_\lambda=(X^TX+\lambda I)^{-1}X^Ty.
$$

1. Compute $w_\lambda$ with `tiny_lambd = 1e-6`. Use `np.linalg.solve` rather than explicitly computing the inverse.
2. Compare its **training accuracy** with the final GD solutions from the small and large initializations.
3. Plot $\|w_t\|_2$ as function of the number of gradient steps $t$ for the two GD runs and add a horizontal line at $\|w_\lambda\|_2$.
4. Plot $\|w_t-w_\lambda \|_2$ as function of the number of gradient steps $t$ for the two GD runs.


What do you observe? Which initialization approaches the tiny-regularization solution more closely?


```python
### your code goes here ###
```

Your answer here: 

```python
# Solution:
tiny_lambd = 1e-6
A = X_small.T@X_small + tiny_lambd*np.eye(d)
w_lambda = np.linalg.solve(A, X_small.T@y_small)

print(f"tiny-ridge train accuracy = {accuracy(X_small, y_small, w_lambda):.3f}")
print(f"small-init GD train accuracy = {accuracy(X_small, y_small, w_small):.3f}")
print(f"large-init GD train accuracy = {accuracy(X_small, y_small, w_large):.3f}")
print()
print(f"||w_lambda|| = {np.linalg.norm(w_lambda):.3f}")
print(f"||w_GD small|| = {np.linalg.norm(w_small):.3f}")
print(f"||w_GD large|| = {np.linalg.norm(w_large):.3f}")


plt.plot(times, norm_small, label="GD: small initialization")
plt.plot(times, norm_large, label="GD: large initialization")
plt.axhline(np.linalg.norm(w_lambda), linestyle="--", label="tiny-ridge solution")
plt.xlabel("t (steps)")
plt.ylabel(r"$||w_t||_2$")
plt.grid(alpha=0.5)
plt.legend()
plt.show()

dist_small = np.linalg.norm(
    np.array(w_small_history) - w_lambda,
    axis=1
)

dist_large = np.linalg.norm(
    np.array(w_large_history) - w_lambda,
    axis=1
)

plt.plot(times, dist_small, label="GD: small initialization")
plt.plot(times, dist_large, label="GD: large initialization")

plt.xlabel("t (steps)")
plt.ylabel(r"$\|w_t - w_\lambda\|_2$")
plt.grid(alpha=0.5)
plt.legend()
plt.show()
```

**Solution** The training accuracies can be almost identical although the weight norms differ. GD from a very small initialization approaches the solution selected by a small $L_2$ regularization, while a large initialization can retain a larger norm. This is the implicit bias of GD in this underdetermined problem.

```python

# Solution:
val_loss_small = np.array([square_loss(X_val, y_val, w) for w in w_small_history[0:500]])
val_acc_small = np.array([accuracy(X_val, y_val, w) for w in w_small_history[0:500]])

best_idx = np.argmin(val_loss_small)
best_t = times[best_idx]
best_w_small = w_small_history[best_idx]

plt.plot(times[0:500], train_loss_small[0:500], label="train loss")
plt.plot(times[0:500], val_loss_small[0:500], label="validation loss")
plt.axvline(best_t, linestyle="--", label="early stopping")
plt.xlabel("t (steps)")
plt.ylabel("square loss")
plt.grid(alpha=0.5)
plt.legend()
plt.show()

plt.plot(times[0:500], train_acc_small[0:500], label="train accuracy")
plt.plot(times[0:500], val_acc_small[0:500], label="validation accuracy")
plt.axvline(best_t, linestyle="--", label="early stopping")
plt.xlabel("t (steps)")
plt.ylabel("accuracy")
plt.grid(alpha=0.5)
plt.legend()
plt.show()

print(f"best validation-loss time: t={best_t}")
print("                 train    validation    test")
print(f"early stopped    {accuracy(X_small,y_small,best_w_small):.3f}       {accuracy(X_val,y_val,best_w_small):.3f}       {accuracy(X_test,y_test,best_w_small):.3f}")
print(f"final GD         {accuracy(X_small,y_small,w_small):.3f}       {accuracy(X_val,y_val,w_small):.3f}       {accuracy(X_test,y_test,w_small):.3f}")

small_early_test_acc = accuracy(X_test, y_test, best_w_small)
small_final_test_acc = accuracy(X_test, y_test, w_small)
```
