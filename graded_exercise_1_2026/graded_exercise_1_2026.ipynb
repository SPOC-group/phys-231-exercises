### Processing the dataset

While the dataset in its original form is intuitive and easy to interpret, we first need to transform it into a numerical representation that we can work with. This transformation is performed through a process called *embedding*. The technical details of how embeddings are constructed are not important for this exercise.

What you should know is that an embedding maps a sentence such as `math and history exams tomorrow` into a matrix in $\mathbb{R}^{L\times D}$. More precisely, the sentence is first split into a sequence of $L$ *tokens*, and each token is then represented by a vector in $\mathbb{R}^D$. By stacking these vectors together, we obtain a matrix of size $L\times D$.

In this representation, $L$ corresponds to the number of tokens in the sentence, while $D$ is the dimension of the vector used to represent each token. The idea is that these vectors provide a numerical representation of the words or tokens, allowing us to manipulate and analyze text using the tools introduced in this course. Since the sentences have different lengths, we fix the sequence length to $L=16$. Sentences longer than $16$ tokens are truncated, while shorter sentences are padded (filled with special tokens) until they reach length $16$.

## Another classification approach: softmax attention

An attention mechanism assigns a weight to each token, allowing the model to give more weight to relevant parts of a tweet. Here, we use the original, uncentered embedding matrices $X_\mu$, rather than the flattened and centered data used for ridge regression. Let $X_{\mu,\ell}\in\mathbb{R}^D$ denote the embedding of token $\ell$ as a column vector; it is stored as row $\ell$ of $X_\mu$.

We define the real-valued prediction
$$\hat y_\mu(k,v)=\frac{1}{\sqrt D}v^T\sum_{\ell=1}^L\sigma_T(\chi_\mu)_\ell X_{\mu,\ell},\qquad \chi_\mu=\frac{X_\mu k}{\sqrt D}\in\mathbb{R}^L,\qquad k,v\in\mathbb{R}^D.$$
The vector $k$ determines the attention weights, while $v$ converts the weighted embedding into a sentiment prediction. The predicted label is $\mathrm{sign}(\hat y_\mu)$.

We train using the sum of squared errors, as in the ridge section:
$$\mathcal L(k,v)=\frac{1}{2}\sum_{\mu=1}^{N}(y_\mu-\hat y_\mu(k,v))^2,\qquad N=n_{\mathrm{train}}.$$
We minimize $\mathcal L$ over $k$ and $v$ using gradient descent.

The softmax with fixed temperature $T>0$ is
$$\sigma_T(\chi_\mu)_\ell=\frac{e^{\chi_{\mu,\ell}/T}}{\sum_{j=1}^L e^{\chi_{\mu,j}/T}}.$$
Use `temperature = 0.7` for training and for plotting the trained model's attention weights. Temperature is a fixed hyperparameter; it is not learned.

**Attention scores and attention weights.**

The quantity $\chi_{\mu,\ell}$ is the raw attention score of token $\ell$. Applying softmax gives its attention weight $\sigma_T(\chi_\mu)_\ell$, which determines how strongly its embedding contributes to the prediction. These weights are nonnegative and sum to one over the $16$ positions, including padding. A larger weight means that the model gives more attention to that token.

For softmax attention, the gradient with respect to $k$ is more involved. We provide the expression to use below:
$$\nabla_k\mathcal L=\frac{1}{T\sqrt D}\sum_{\mu=1}^{N}(\hat y_\mu(k,v)-y_\mu)\sum_{\ell=1}^{L}\sigma_T(\chi_\mu)_\ell\left[\frac{X_{\mu,\ell}^T v}{\sqrt D}-\hat y_\mu(k,v)\right]X_{\mu,\ell}.$$

We provide implementations of the prediction, loss, and gradients below. Study how the code is structured, but do not change it.

```python
def attentionSoftmax(X, k, v, temperature=0.7):
    dim = X.shape[-1]
    weights = softmax(X @ k / np.sqrt(dim), temperature)
    return np.sum(weights * (X @ v), axis=-1) / np.sqrt(dim)
def loss(y, y_pred):
    return 0.5 * np.sum((y - y_pred) ** 2)
def gradLossSoftmax(X, y, k, v, temperature=0.7): 
    dim = X.shape[-1]
    weights = softmax(X @ k / np.sqrt(dim), temperature)
    values = X @ v / np.sqrt(dim)
    prediction = np.sum(weights * values, axis=-1)
    residual = prediction - y
    grad_k = np.einsum("n,nl,nld->d", residual, weights * (values - prediction[:, None]), X)
    grad_k /= temperature * np.sqrt(dim)
    grad_v = np.einsum("n,nl,nld->d", residual, weights, X) / np.sqrt(dim)
    return grad_k, grad_v
```
